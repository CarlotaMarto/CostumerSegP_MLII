<h2 style="margin:0;"><b>1) Imports and data loading</b></h2>
</div>

```python
%load_ext autoreload
%autoreload 2

import os, sys, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(os.path.abspath('.'))
warnings.filterwarnings('ignore')

import utils                # professor's helpers (plot_dendrogram, ...)
import utils_clustering as uc

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
pd.set_option('display.max_columns', None)
```

```python
DATA_DIR = '../datasets'
regular = pd.read_csv(f'{DATA_DIR}/info_clustering_unscaled.csv', index_col='customer_id')
outliers = pd.read_csv(f'{DATA_DIR}/outlier_dataset.csv', index_col='customer_id')
print('Regular:', regular.shape, '| Outliers:', outliers.shape,
      '| Total:', len(regular) + len(outliers))
regular.head()
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #0047AB;">
<h2 style="margin:0;"><b>2) The menu — candidate feature sets and scalers</b></h2>
<p style="margin:6px 0 0 0;">These are the options the diagnostics will score. You can edit <code>build_candidate_feature_sets</code> in <code>utils_clustering.py</code> to add or remove sets. Each set is <code>(columns, log_absolute_spend?)</code>.</p>
</div>

<div style="background-color:#F7FBFF; padding:15px; border-radius:10px;">
Why some columns repeat information: absolute <code>lifetime_spend_*</code>, share <code>pct_spend_*</code> and <code>log_total_spend</code> all describe spend; <code>tenure</code> = current year − <code>year_first_transaction</code>; <code>total_children</code> = <code>kids_home + teens_home</code>; <code>typical_hour</code> is replaced by its cyclic <code>sin/cos</code>. Avoid putting two views of the same thing in one feature set. Identity/geo (<code>customer_id, latitude, longitude, is_male, customer_loyalty_flag</code>) are best left out of the distance and used for profiling.
</div>

```python
candidate_sets = uc.build_candidate_feature_sets(regular)
for name, (cols, logabs) in candidate_sets.items():
    print(f'{name:32s} | {len(cols):2d} cols | log_abs_spend={logabs}')

scaler_options = ['Standard', 'MinMax', 'Robust', 'None']
print('\nScalers available:', scaler_options)
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #0047AB;">
<h2 style="margin:0;"><b>3) Diagnostics A — feature set × k silhouette grid</b></h2>
<p style="margin:6px 0 0 0;">A single heatmap of mean silhouette for every feature set across k, under one scaler. Higher = better separated. Re-run with a different <code>scaler_name</code> to see how sensitive things are.</p>
</div>

```python
grid = uc.silhouette_grid(regular, candidate_sets, k_range=range(2, 11),
                          scaler_name='Robust')
grid_pivot = uc.plot_silhouette_grid(grid, title='Silhouette grid (scaler = Standard)')
grid_pivot
```

<div style="background-color:#F7FBFF; padding:15px; border-radius:10px;">
<b>How to read it.</b> Look for a row (feature set) that stays high, and a column (k) where it peaks. A high silhouette means tight, well-separated clusters — but a very high score at k = 2 usually just means "everyone vs a tiny group", which is rarely useful for marketing. Balance separation against how many distinct, targetable segments you actually want.
</div>

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #0047AB;">
<h2 style="margin:0;"><b>4) Diagnostics B — scaler comparison, elbow and dendrogram</b></h2>
<p style="margin:6px 0 0 0;">Pick ONE feature set to inspect more closely (set <code>inspect_set</code>), then compare scalers, view the inertia elbow, and view a Ward dendrogram. None of this commits you to anything.</p>
</div>

```python
# Point this at any feature set name from Section 2 to inspect it.
inspect_set = 'spend + promo no groceries'
inspect_cols, inspect_logabs = candidate_sets[inspect_set]
print(f'Inspecting: {inspect_set}  ({len(inspect_cols)} cols, log_abs={inspect_logabs})')
```

### 4.1 — Scaler comparison (silhouette vs k)

```python
scaler_scores = uc.compare_scalers(regular, inspect_cols, k_range=range(2, 11))
uc.plot_scaler_comparison(scaler_scores)
scaler_scores.pivot(index='k', columns='scaler', values='silhouette')
```

<div style="background-color:#F7FBFF; padding:15px; border-radius:10px;">
<b>A note on scaling the share-of-wallet features.</b> The <code>pct_spend_*</code> columns are already proportions in [0, 1] that sum to ~1 per customer. Re-applying <b>MinMax</b> rescales each category by its own observed min/max, which stretches rare categories and distorts the comparability between categories — so even if it scores marginally higher on silhouette, it is hard to justify on this data. <b>StandardScaler</b> (z-score, the choice in the Week-6 walkthrough) or even no scaling are more defensible here. We default to Standard for the share-of-wallet set.
</div>

### 4.2 — Inertia elbow

```python
X_inspect = uc.apply_feature_pipeline(regular, inspect_cols, inspect_logabs,
                                      uc.get_scaler('Robust'), fit=True)
k_values, inertia = uc.kmeans_elbow(X_inspect, range(1, 11))
uc.plot_elbow(k_values, inertia)
```

### 4.3 — Ward dendrogram (on a sample)

<div style="background-color:#F7FBFF; padding:15px; border-radius:10px;">
Agglomerative clustering is O(n²) in memory, so we draw the dendrogram on a random sample. Reading where long vertical links get cut suggests candidate values of k.
</div>

```python
_, dendro_model = uc.fit_dendrogram_model(X_inspect, sample_size=3000, linkage='ward')
fig, ax = plt.subplots(figsize=(12, 5))
plt.title(f'Ward dendrogram (sample) — {inspect_set}')
utils.plot_dendrogram(dendro_model, truncate_mode='level', p=5)
plt.xlabel('Sample of customers'); plt.ylabel('Distance')
plt.show()
```

<div style="background-color:#FFF6E5; padding:20px; border-radius:10px; border-left:8px solid #E08600;">
<h2 style="margin:0;"><b>5) YOUR CHOICES</b></h2>
<p style="margin:6px 0 0 0;">This is the only cell that makes a decision. Set the three values below from the menus above, then run everything that follows. Change them and re-run as often as you want.</p>
</div>

```python
# ===========================  YOUR DECISIONS  ===========================
# 1) FEATURE_SET : any key printed in Section 2
# 2) SCALER      : 'Standard' | 'MinMax' | 'Robust' | 'None' (sem escalar)
# 3) K           : the number of clusters you decide on
# ------------------------------------------------------------------------
FEATURE_SET = 'spend + promo no groceries' # <- change me
SCALER      = 'Robust'         # <- change me
K           = 8                   # <- change me
# ========================================================================

distance_cols, LOGABS = candidate_sets[FEATURE_SET]
profiling_cols = uc.get_profiling_features(regular, distance_cols)
scaler = uc.get_scaler(SCALER)
X = uc.apply_feature_pipeline(regular, distance_cols, LOGABS, scaler, fit=True)

print(f'Feature set : {FEATURE_SET}  ({len(distance_cols)} cols, log_abs={LOGABS})')
print(f'Scaler      : {SCALER}')
print(f'k           : {K}')
print(f'Matrix      : {X.shape}')
```

<b>Decision log (optional but recommended for the report).</b> In a sentence or two, write why you picked this feature set, scaler and k — e.g. which diagnostic convinced you and what trade-off you accepted. Double-click to edit:<br><br><i>I chose &lt;…&gt; because &lt;…&gt;. I accepted &lt;…&gt; in exchange for &lt;…&gt;.</i>
</div>

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #0047AB;">
<h2 style="margin:0;"><b>6) Fit and validate your choice</b></h2>
<p style="margin:6px 0 0 0;">Fit K-Means with your settings and inspect cluster sizes, the silhouette plot, and PCA / UMAP projections. If you do not like what you see, go back to Section 5.</p>
</div>

```python
kmeans = uc.fit_kmeans(X, K)
regular['cluster'] = kmeans.labels_
uc.plot_cluster_sizes(regular, 'cluster')
uc.cluster_sizes(regular, 'cluster')
```

### 6.1 — Silhouette plot

```python
avg_sil = uc.plot_silhouette_blades(X, kmeans.labels_, title=f'Silhouette — KMeans (k={K})')
print('Average silhouette:', round(avg_sil, 3))
```

### 6.2 — PCA and UMAP projections

<div style="background-color:#F7FBFF; padding:15px; border-radius:10px;">
UMAP needs <code>umap-learn</code> (<code>pip install umap-learn</code>); if it is not installed the helper falls back to t-SNE automatically. We subsample to 8k points so the plots are fast.
</div>

```python
X_emb, lab_emb = uc.subsample(X, kmeans.labels_, n=8000)
pca_emb = uc.embed_pca(X_emb)
uc.plot_embedding(pca_emb, lab_emb, title='PCA - KMeans Clusters', method_name='PCA')
```

```python
umap_emb, method = uc.embed_umap(X_emb)
uc.plot_embedding(umap_emb, lab_emb, title='UMAP - KMeans Clusters', method_name=method)
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #0047AB;">
<h2 style="margin:0;"><b>7) Hierarchical cross-check</b></h2>
<p style="margin:6px 0 0 0;">Compare your K-Means labels with a Ward solution at the same k on a sample. Strong agreement (most mass on one Ward column per K-Means row) means the structure is real.</p>
</div>

```python
sample_idx, ward_labels = uc.fit_hierarchical_sample(X, K, sample_size=5000, linkage='ward')
uc.compare_solutions(kmeans.labels_[sample_idx], ward_labels, name_a='KMeans', name_b='Ward')
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #0047AB;">
<h2 style="margin:0;"><b>8) Segment profiling</b></h2>
<p style="margin:6px 0 0 0;">Interpret each cluster against the OVERALL mean (the anchor). These tables feed the report; they use ALL variables, not just the ones that drove the distance.</p>
</div>

```python
spend_share = [c for c in regular.columns if c.startswith('lifetime')]
share_profile = uc.profile_clusters(regular, 'cluster', spend_share, as_percent=True)
uc.plot_profile_heatmap(share_profile, 'Share of wallet by cluster (%)')
share_profile
```

```python
abs_cols = [c for c in regular.columns if c.startswith('lifetime_spend_')]
uc.profile_clusters(regular, 'cluster', abs_cols)
```

```python
key_cols = [c for c in ['log_total_spend','customer_age','education_level','total_children',
                        'tenure','percentage_of_products_bought_promotion','number_complaints',
                        'distinct_stores_visited','customer_loyalty_flag','is_male']
            if c in regular.columns]
uc.profile_clusters(regular, 'cluster', key_cols)
```

```python
if {'latitude','longitude'}.issubset(regular.columns):
    plt.figure(figsize=(8, 7))
    sns.scatterplot(data=regular, x='longitude', y='latitude',
                    hue='cluster', palette='tab10', s=8, alpha=0.5, legend='full')
    plt.title('Customer locations by segment (profiling only)')
    plt.show()
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #0047AB;">
<h2 style="margin:0;"><b>9) Re-attach outliers and export</b></h2>
<p style="margin:6px 0 0 0;">Project the held-aside outliers through the SAME pipeline (same columns, same log step, same fitted scaler) and assign each to its nearest centroid, then export one cluster per customer. The assert guarantees every customer is present.</p>
</div>

```python
X_out = uc.apply_feature_pipeline(outliers, distance_cols, LOGABS, scaler, fit=False)
outlier_labels = kmeans.predict(X_out)
print('Outliers assigned:', len(outlier_labels))

segments = uc.build_full_assignment(regular, kmeans.labels_, outliers, outlier_labels,
                                    label_col='cluster', id_name='customer_id')
assert segments['customer_id'].nunique() == len(regular) + len(outliers), 'Missing customers!'
print('Customers in final file:', segments['customer_id'].nunique())
print(segments['cluster'].value_counts().sort_index())
```

```python
segments.to_csv(f'{DATA_DIR}/customer_segments.csv', index=False)
uc.cluster_sizes(regular, 'cluster').to_csv(f'{DATA_DIR}/segment_summary.csv')
share_profile.to_csv(f'{DATA_DIR}/segment_share_profile.csv')
print('Saved customer_segments.csv (+ summary and share profile) to', DATA_DIR)
```
